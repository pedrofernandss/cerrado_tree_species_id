```python
import shutil
import re
from pathlib import Path
```

```python
def extract_id(nome_arquivo):
    match = re.search(r"(DJI_\d{14}_\d{4})", nome_arquivo)
    if match:
        return match.group(1) 
    return None
```

```python
path_labels_source = Path("/datasets/projeto_cerrado/fotos-rotuladas/rgb/labels")
path_images_source = Path("/datasets/projeto_cerrado/2024-08-28/rgb-ndre-imgs")
path_output = Path("/datasets/projeto_cerrado/fotos-rotuladas/rgb-ndre")
```

```python
(path_output / "images").mkdir(parents=True, exist_ok=True)
(path_output / "labels").mkdir(parents=True, exist_ok=True)
```

```python
dict_imagens = {}
lista_imgs = list(path_images_source.glob("*.[tT][iI][fF]")) 

for img in lista_imgs:
    id_img = extract_id(img.name)
    if id_img:
        dict_imagens[id_img] = img
    else:
        print(f"Image without ID: {img.name}")

labels_list = list(path_labels_source.glob("*.txt"))

matches = 0
erros = 0

for lbl in labels_list:
    id_lbl = extract_id(lbl.name)
    
    if id_lbl and id_lbl in dict_imagens:
        img_path = dict_imagens[id_lbl]
        
        try:
            shutil.copy(img_path, path_output / "images" / img_path.name)

            novo_nome_label = img_path.stem + ".txt"
            shutil.copy(lbl, path_output / "labels" / novo_nome_label)
            
            matches += 1
            
        except Exception as e:
            print(f"Erro to split {id_lbl}: {e}")
            erros += 1

print(f"Total imagens in original folder Fonte: {len(lista_imgs)}")
print(f"Total labels in original folder:  {len(labels_list)}")
print(f"Matches:         {matches}")
```
